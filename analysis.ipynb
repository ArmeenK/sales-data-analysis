```python
df[df["Region"] == "West"]
#pandas reads it as: "for every row, if the value here is True, keep the whole row; if False, throw it away." 
#So the outer df[...] is a lookup into your original dataframe, using the inner True/False list as a mask/filter.
```

```python
df.isnull().sum()
```

```python
df["Postal Code"] = df["Postal Code"].fillna(0)
df.isnull().sum()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
df.dtypes
```

```python
df["Order Processing"] = df["Ship Date"] - df["Order Date"]
df["Order Processing"]
```

```python
import numpy as np
df["Sales"].values #converts dataframe column into nparray
np.std(df["Sales"]) #standard deviation
```

```python
df.groupby("Category")["Sales"].sum().plot(kind="bar", color="pink")
plt.title("Total Sales by Category")
plt.ylabel("Total Sales ($)")
plt.xlabel("Category")
plt.show()
```

```python
df["Order Month"]=df['Order Date'].dt.to_period('M')
```

```python
df.groupby("Order Month")["Sales"].sum()
```

```python
df.groupby("Order Month")["Sales"].sum().plot(kind="line", color="purple")
plt.title("Total Sales Over Months")
plt.xlabel("Months & Years")
plt.ylabel("Total Sales")
plt.show()
```
